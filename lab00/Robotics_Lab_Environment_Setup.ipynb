### Step 2 — Verify Core Libraries

Run the following cell.

````python
import numpy as np
import scipy
import matplotlib
import cvxpy

print("Robotics environment check")
print("---------------------------")
print("NumPy      :", np.__version__)
print("SciPy      :", scipy.__version__)
print("Matplotlib :", matplotlib.__version__)
print("CVXPY      :", cvxpy.__version__)

print("\nCore robotics environment: READY")
````

### Step 3 — Verify Visualization

A robotics environment must support visualization of robot states,
trajectories, sensor measurements, and simulation results.

````python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 10, 200)
y = np.sin(x)

plt.figure(figsize=(8, 4))
plt.plot(x, y)
plt.xlabel("x")
plt.ylabel("sin(x)")
plt.title("Robotics Visualization Check")
plt.grid(True)
plt.show()
````
